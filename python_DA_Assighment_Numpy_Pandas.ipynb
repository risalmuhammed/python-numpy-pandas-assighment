```python
# Python DA Assignment 1: Data Analysis using NumPy and Pandas

import numpy as np
import pandas as pd

print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)
```

```python
# 1. Create a 1D NumPy Array for Week 1

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:", temperatures_w1)
```

```python
# 2. Inspect Array Properties

print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
# 3. Convert Celsius to Fahrenheit

temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Celsius:", temperatures_w1)
print("Temperatures in Fahrenheit:", temperatures_f)
```

```python
# Find Maximum, Minimum, and Mean Temperatures

print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

```python
# 4. Array Slicing and Indexing

# Temperatures for the first three days
print("First three days:", temperatures_w1[:3])
```

```python
# Temperatures for the weekend (last two days)
print("Weekend temperatures:", temperatures_w1[-2:])
```

```python
# Temperatures for the middle three days
print("Middle three days:", temperatures_w1[2:5])
```

```python
# 5. Create a 2D NumPy Array for Two Weeks

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Temperatures for Two Weeks:")
print(temperatures)
```

```python
# Inspect the 2D Array

print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)
```

```python
# Extract temperatures for each week

print("Week 1:", temperatures[0])
print("Week 2:", temperatures[1])

# Extract weekend temperatures (last two days) for both weeks

print("Week 1 Weekend:", temperatures[0, -2:])
print("Week 2 Weekend:", temperatures[1, -2:])
```

```python
# STEP 7: PANDAS SERIES

# 1. Create a Pandas Series
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Student Marks:")
print(marks)
```

```python
# STEP 8: INDEXING AND SLICING

# 1. Access the first rank student's mark using integer position
print("First rank mark:", marks.iloc[0])

# 2. Retrieve the top 3 ranks using index labels
print("\nTop 3 rank students:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# 3. Retrieve the 3rd rank student's mark using iloc
print("\nThird rank mark:", marks.iloc[2])

# 4. Filter students whose marks are greater than 90
print("\nStudents scoring above 90:")
print(marks[marks > 90])
```

```python
# STEP 9: MANIPULATING PANDAS SERIES

# 1. Modify the mark of the Rank1 student to 100
marks.loc['Rank1'] = 100
print("Updated marks:")
print(marks)

# 2. Remove the entry corresponding to the last rank
marks = marks.drop('Rank5')
print("\nMarks after removing Rank5:")
print(marks)

# 3. Compute CGPA by dividing each mark by 10
cgpa = marks / 10
print("\nCGPA:")
print(cgpa)
```

```python
# STEP 10: CREATING A PANDAS DATAFRAME

# Create the transactions DataFrame
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print("Transactions DataFrame:")
print(transactions)
```

```python
# STEP 11: DATA EXPLORATION

# 1. Display the first 5 rows
print("First 5 rows:")
print(transactions.head())

# 2. Display the last 5 rows
print("\nLast 5 rows:")
print(transactions.tail())

# 3. Display the shape
print("\nShape:")
print(transactions.shape)

# 4. Display column names
print("\nColumn Names:")
print(transactions.columns)

# 5. Display data types
print("\nData Types:")
print(transactions.dtypes)

# 6. Display basic information
print("\nDataFrame Information:")
transactions.info()

# 7. Display only ProductCategory and Amount
print("\nProductCategory and Amount:")
print(transactions[['ProductCategory', 'Amount']])

# 8. Retrieve the last 3 columns using iloc
print("\nLast 3 Columns:")
print(transactions.iloc[:, -3:])
```

```python
# STEP 12: FILTERING AND ANALYZING DATA

# 1. Filter North region transactions with Amount greater than 200
print("North Region Transactions with Amount > 200:")
print(transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
])

# 2. Count transactions in each product category
print("\nProduct Category Counts:")
print(transactions['ProductCategory'].value_counts())

# 3. Find unique regions
print("\nUnique Regions:")
print(transactions['Region'].unique())

# 4. Calculate average transaction amount for each region
print("\nAverage Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
```

```python
# STEP 13: MANIPULATING THE DATAFRAME

# 1. Update the Amount for TransactionID 102 to 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print("After updating TransactionID 102:")
print(transactions)

# 2. Add a Discount column (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10

print("\nAfter adding the Discount column:")
print(transactions)

# 3. Remove the transaction with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]

print("\nAfter removing TransactionID 109:")
print(transactions)

# 4. Delete the Discount column
transactions = transactions.drop(columns=['Discount'])

print("\nAfter deleting the Discount column:")
print(transactions)
```

```python
# STEP 14: FINAL CHECK

print("Final Transactions DataFrame:")
print(transactions)

print("\nDataFrame Shape:", transactions.shape)

print("\nMissing Values:")
print(transactions.isnull().sum())

print("\nDuplicate Rows:", transactions.duplicated().sum())

print("\nTotal Transaction Amount:", transactions['Amount'].sum())

print("\nAverage Transaction Amount:", transactions['Amount'].mean())
```
